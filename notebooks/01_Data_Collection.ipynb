Module 2 – Data Cleaning & Preprocessing

Instead of deleting everything with missing values (which many projects do), we'll follow a production-ready strategy.

```python
# Check duplicate rows

print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Convert date column into datetime format

df["date"] = pd.to_datetime(df["date"])

print(df["date"].dtype)
```

```python
missing = (
    df.isnull()
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

missing.head(20)
```

```python
# Columns having more than 90% missing values

high_missing = missing[missing > 90]

high_missing
```

```python
# Drop columns having more than 90% missing values

df = df.drop(columns=high_missing.index)

print("Remaining Columns:", df.shape[1])
```

```python
# Verify removed columns

set(high_missing.index).intersection(df.columns)
```

```python
print(df.shape)
```

```python
# Check country-date structure

print("Countries:", df["location"].nunique())
print("Date range:", df["date"].min(), "to", df["date"].max())

print("\nDuplicate country-date combinations:")
print(df.duplicated(subset=["location", "date"]).sum())
```

```python
# Sort data by country and date

df = df.sort_values(
    by=["location", "date"]
).reset_index(drop=True)

print(df[["location", "date"]].head(10))
```

```python
# Check missing values in key analysis variables

key_columns = [
    "total_cases",
    "new_cases",
    "total_deaths",
    "new_deaths",
    "total_tests",
    "new_tests",
    "total_vaccinations",
    "new_vaccinations",
    "people_fully_vaccinated"
]

missing_key = df[key_columns].isna().sum().sort_values(ascending=False)

print(missing_key)
```

```python
# Missing percentage for key variables

missing_key_pct = (
    df[key_columns]
    .isna()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
    .round(2)
)

print(missing_key_pct)
```

```python
# Fill missing case and death values with 0

case_death_columns = [
    "total_cases",
    "new_cases",
    "total_deaths",
    "new_deaths"
]

df[case_death_columns] = df[case_death_columns].fillna(0)

print("Missing values after filling:")
print(df[case_death_columns].isna().sum())
```
